#Cleaning & Preping gameinfo.csv

```python
gameinfo.columns
```

```python
game_info_cols_of_interest = ["gid","visteam", "hometeam", "site","number","daynight","attendance",
                             "precip","sky","temp","winddir","windspeed", "suspend", "forfeit", "gametype", "season",
                            "vruns","hruns", "wteam", "lteam"]

gameinfo2 = gameinfo[game_info_cols_of_interest]

"""
Features with null vals 

number : "nan"

#0-3 means at least 1 game was played
null count : 12316 

daynight : "nan", "unknown"
unknown count: 5
null count: 1269

attendance : 
null count : 2154 missing vals
#solution : groupby site, use average attendance

precip : "unknown", "nan" 
unknown count: 162013
null count: 3658

sky : "unknown" , "nan"
unknown count: 114066
null count: 3663

temp : "unknown", "nan" *note there is mixed datatype in this col
unknown count: 78
null count: 3662

winddir : "unknown" , "nan"
unknown count: 141667
null count: 3664

windspeed : "nan", "unknown" *not there is mixed datatype in this col
unknown count: 78
null count: 3665

suspend : "nan indicates game was not suspended" *we are only using rows that are nan for this col
unknown count: 0
null count: 214747 #Possible data enteries of not suspended games

forfeit : "nan indicates no forfeit"
unknown count: 0
null count: 214900 #Possible data enteries of not forfeited games
"""
```

```python
for col in game_info_cols_of_interest:
    print(col,gameinfo2[col].unique(),"\n\n")     
```

```python
col = "forfeit"
null_count = gameinfo2[col].isnull().sum() #nan

unknown_count = gameinfo2.loc[gameinfo2[col] == "unknown"]
print("unknown count:" , unknown_count[col].count())
print("null count:", null_count)
```

```python
gameinfo2.shape
```

```python
#Filter gameinfo2 for rows suspend ,  forfeit = nan

gameinfo3 = gameinfo2.loc[gameinfo2["suspend"].isnull()]
gameinfo4 = gameinfo3.loc[gameinfo3["forfeit"].isnull()] #14,704 entries (None forfeit, None suspended games)

gameinfo5 = gameinfo4.loc[gameinfo4["gametype"] == "regular"] #filter for regular season games : 214,682 entries
gameinfo5.shape
```

```python
#Features to be dropped due to large unknown/nan count 
drop_cols = ["precip" , "sky" , "winddir", "forfeit", "suspend", "gametype"] #dropping gametype,suspend and forfeit post filtering
gameinfo6 = gameinfo5.drop(drop_cols, axis=1)
```

```python
gameinfo6.columns
```

```python
# "number" 
gameinfo6.loc[gameinfo6["number"].isnull()].head() 
#12,316 missing values from season 1901 - 1911

#will replace NaN with 0, Reason : Domain knowledge, baseball teams rarely play multiple games in one day in the pros
gameinfo6["number"] = gameinfo6["number"].fillna(0)
```

```python
# "daynight" 

gameinfo6.loc[(gameinfo["daynight"] == "unknown") | (gameinfo["daynight"].isnull())]
#1262 rows with eith NaN or "unknown" 1921 - 1949
```

```python
# attendance
sites = gameinfo6["site"].unique()
gameinfo6["attendance"] = gameinfo6["attendance"].fillna(0)
#attendance columns mixes text and numbers, dtype = object

col = gameinfo6.columns.get_loc("attendance")
for i in range(gameinfo6.shape[0]):
    try:
        gameinfo6.iat[i, col] = int(gameinfo6.iat[i, col])  # Use iat for faster access
    except (ValueError, TypeError):
        gameinfo6.iat[i, col] = 0  # Set to 0 if conversion fails

gameinfo6["attendance"] = gameinfo6["attendance"].astype(int) 
gameinfo6["attendance"].dtype
```

```python
gameinfo6.loc[gameinfo6["attendance"] == 0 ]
#Missing data for attendance is randomly missing 1901-2024
```

```python
# temp
gameinfo6["temp"] = gameinfo6["temp"].fillna(0)
#temp columns mixes text and numbers, dtype = object

col = gameinfo6.columns.get_loc("temp")
for i in range(gameinfo6.shape[0]):
    try:
        gameinfo6.iat[i, col] = float(gameinfo6.iat[i, col])  # Use iat for faster access
    except (ValueError, TypeError):
        gameinfo6.iat[i, col] = 0  # Set to 0 if conversion fails

gameinfo6["temp"] = gameinfo6["temp"].astype(float) 
gameinfo6["temp"].dtype
```

```python
# windspeed
feature = "windspeed"
gameinfo6[feature] = gameinfo6[feature].fillna(0)


col = gameinfo6.columns.get_loc(feature)
for i in range(gameinfo6.shape[0]):
    try:
        gameinfo6.iat[i, col] = float(gameinfo6.iat[i, col])  # Use iat for faster access
    except (ValueError, TypeError):
        gameinfo6.iat[i, col] = 0  # Set to 0 if conversion fails

gameinfo6[feature] = gameinfo6[feature].astype(float) 
gameinfo6[feature].dtype
```
