```python
import pandas as pd
```

```python
gameinfo = pd.read_csv("../CleanData/gameinfo_4_25_25.csv")
team_salaries = pd.read_csv("../CleanData/team_salaries_4_26_25.csv")
team_stats = pd.read_csv("../CleanData/team_stats_04_27_25.csv")
```

```python
gameinfo = gameinfo.loc[gameinfo["season"] > 1984]
gameinfo
```

```python
team_salaries = team_salaries.loc[team_salaries["yearID"] < 2025]
team_salaries
```

```python
team_stats = team_stats.drop("Unnamed: 0",axis = 1)
team_stats
```

#Randomly sampling to equally represent Home and Visting Team

```python
import numpy as np

rows_count = gameinfo.shape[0]
samples = np.random.binomial(1, 0.5, rows_count)

samples #Home Team = 0 , Visting Team = 1
```

```python
teams_of_interest = []

home_col_index = gameinfo.columns.get_loc("hometeam")
visit_col_index = gameinfo.columns.get_loc("visteam")

for i in range(len(samples)):
    if samples[i] == 0:
        team = gameinfo.iloc[i, home_col_index]
    else:
        team = gameinfo.iloc[i,visit_col_index]
    teams_of_interest.append(team)

teams_of_interest
```

```python
#Two new columns added
gameinfo.loc[:, "team"] = teams_of_interest
gameinfo.loc[:, "Home_Away"] = samples #Home Team = 0 , Visting Team = 1
gameinfo
```

```python
#Difference in score "team" - opponent
home_score_col_index = gameinfo.columns.get_loc("hruns")
visit_score_col_index = gameinfo.columns.get_loc("vruns")

score_diff = []

for i in range(len(samples)):
    hruns = gameinfo.iloc[i, home_score_col_index]
    vruns = gameinfo.iloc[i, visit_score_col_index]
    
    if samples[i] == 0:
        score = hruns - vruns
    else:
        score = vruns - hruns
    score_diff.append(score)

score_diff
```

```python
gameinfo.loc[:, "score_diff"] = score_diff
gameinfo
```

```python
#merged with teamstats data
merge_cols = ["gid", "team"]

merged_1 = pd.merge(gameinfo, team_stats, on=merge_cols, how='left')
merged_1
```

```python
#Merged with salary data
merged_2 = pd.merge(
    merged_1, 
    team_salaries, 
    left_on=["season", "team"], 
    right_on=["yearID", "teamID"], 
    how="left"
)
merged_2
```

```python
col_dtype_conv = ["gid","visteam","hometeam","site","daynight","wteam", "lteam","team", "teamID"]
merged_2.loc[:,col_dtype_conv] = merged_2.loc[:,col_dtype_conv].astype("string")
```

```python
merged_2['daynight'] = merged_2['daynight'].map({'day': 1, 'night': 0})
```

#Imputing Data For Missing Values

```python
#check 0 val for attendance , temp, windspeed , for these cols NaNs were replaced with 0s.
#Original data did not have 0 vals in these cols
```

```python
averages = merged_2.groupby("site")[["attendance", "temp", "windspeed"]].mean().reset_index()
averages
```

```python
attendance_col_index = merged_2.columns.get_loc("attendance")
temp_col_index = merged_2.columns.get_loc("temp")
windsp_col_index = merged_2.columns.get_loc("windspeed")
site_col_index = merged_2.columns.get_loc("site")

for i in range(merged_2.shape[0]):
    attendance = merged_2.iloc[i,attendance_col_index]
    temp = merged_2.iloc[i,temp_col_index]
    windspeed = merged_2.iloc[i,windsp_col_index]
    site = merged_2.iloc[i,site_col_index]

    if attendance == 0:
        merged_2.iloc[i,attendance_col_index] = int(averages.loc[averages["site"] == site,"attendance"].values[0])
    if windspeed == 0:
        merged_2.iloc[i,windsp_col_index] = averages.loc[averages["site"] == site,"windspeed"].values[0]
    if temp == 0:
        merged_2.iloc[i,temp_col_index] = averages.loc[averages["site"] == site,"temp"].values[0]
```
