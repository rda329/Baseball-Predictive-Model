```python
import pandas as pd
```

```python
team_stats = pd.read_csv("Data/teamstats.csv")
```

# Cleaning & Preping team_stats.csv

```python
print(team_stats.columns)
print(team_stats.shape)
```

```python
team_stats_cols_of_interest = ["gid", "team", "lob", "b_pa", "b_d", "b_t", "b_hr", "b_sb", "b_roe", "p_cs", "p_pb", "d_e", "d_sb", "date"]

team_stats2 = team_stats[team_stats_cols_of_interest]
team_stats2['date'] = pd.to_datetime(team_stats2['date'], format='%Y%m%d')
```

### rename columns to meaningful names

```python
# rename columns to meaningful names
original_column_names = ["gid", "team", "lob", "b_pa", "b_d", "b_t", "b_hr", "b_sb", "b_roe", "p_cs", "p_pb", "d_e", "d_sb", "date"]
new_column_names = ["gid", "team", "left_on_base", "bat_plate_appear", "bat_doubles", "bat_triples", "bat_home_runs", "bat_stolen_bases", "bat_base_reaches_w_error", "caught_stealing_allowed", "passed_balls_allowed", "errors_committed", "stolen_bases", "date"]

team_stats2 = team_stats2.rename(columns=dict(zip(original_column_names, new_column_names)))
team_stats2.isna().sum()
```

 print initial unique values per column

```python
def print_unique_values(df, ignore_cols=None):
    # unique values for each column
    if ignore_cols is None:
        ignore_cols = ["gid", "team", "date"]

    for col in df.columns:
        if col in ignore_cols:
            # ignore these columns as they dont have missing values
            continue
        col_unique_values = df[col].unique()
        col_na_count = int(df[col].isna().sum())
        print(f"{col}: {col_unique_values} \n{col_na_count=} \n")

print_unique_values(team_stats2)
```

```python
# check column values bat_plate_appear: NA values

bat_plate_appear_na_rows = team_stats2[team_stats2["bat_plate_appear"].isna()]
print(bat_plate_appear_na_rows.shape)
bat_plate_appear_na_rows.head(5)
```

```python
# Features to be dropped due to large nan count
drop_cols = ["bat_plate_appear", "bat_base_reaches_w_error", "caught_stealing_allowed"]
team_stats3_drop_too_many_nan = team_stats2.drop(drop_cols, axis=1)
team_stats3_drop_too_many_nan.shape
```

```python
team_stats3_drop_too_many_nan.columns
```

```python
team_stats3_drop_too_many_nan.loc[team_stats3_drop_too_many_nan["left_on_base"].isna()].head()
```

```python
# drop rows with multiple NaN values
team_stats4_clean = team_stats3_drop_too_many_nan.dropna(thresh=5)
team_stats4_clean.shape
```

```python
team_stats4_clean.loc[team_stats4_clean["left_on_base"].isna()].head()
```

```python
print_unique_values(team_stats4_clean)
```

#### clean columns with considerable fewer NaN values than others

```python
team_stats5_clean = team_stats4_clean.dropna(subset=['left_on_base', 'bat_doubles', 'bat_triples', 'bat_home_runs', 'bat_stolen_bases', 'errors_committed'])
```

```python
# unique values after cleaning
print_unique_values(team_stats5_clean)
total_rows = team_stats5_clean.shape[0]
print(f"Total rows: {total_rows}")
```
